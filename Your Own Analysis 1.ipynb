```python
import gzip
from Bio import SeqIO

file_path = "/home/dgh84/yoa/SRR2057647.fastq.gz"

# 처음 3개의 read 확인
with gzip.open(file_path, "rt") as handle:
    for i, record in enumerate(SeqIO.parse(handle, "fastq")):
        print(f"[{i+1}] ID: {record.id}")
        print(f"Seq : {record.seq}")
        print(f"Qual: {record.letter_annotations['phred_quality']}")
        print("-" * 50)
        if i == 2:
            break
```

```python
count = 0
with gzip.open(file_path, "rt") as handle:
    for _ in SeqIO.parse(handle, "fastq"):
        count += 1
print(f"Total reads: {count}")
```
